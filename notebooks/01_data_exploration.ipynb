## Sensor Trends
Let's visualize the trends of sensors for a specific engine.

```
engine_1 = train_df[train_df['unit_id'] == 1]
plt.figure(figsize=(10,4))
plt.plot(engine_1['cycle'], engine_1['sensor_2'])
plt.title('Sensor 2 vs Cycle (Unit 1)')
plt.xlabel('Cycle')
plt.ylabel('Sensor 2 Value')
plt.show()
```
